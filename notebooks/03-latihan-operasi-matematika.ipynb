### Soal 1: Luas Persegi

Hitung luas persegi dengan sisi $5\ cm$.

Rumus: $L = s \times s = s^2$

````python
# Tulis kode Anda
sisi = 5
luas = sisi ** 2
print(f"Luas persegi = {luas} cm²")
````

### Soal 2: Volume Kubus

Hitung volume kubus dengan rusuk $4\ cm$.

Rumus: $V = r^3$

````python
# Tulis kode Anda
rusuk = 4
volume = rusuk ** 3
print(f"Volume kubus = {volume} cm³")
````

### Soal 3: Akar Kuadrat

Hitung $\sqrt{144}$ dan $\sqrt{225}$ menggunakan `math.sqrt()`.

````python
# Tulis kode Anda
import math

print(f"Akar dari 144 = {math.sqrt(144)}")
print(f"Akar dari 225 = {math.sqrt(225)}")
````

### Soal 4: Pythagoras

Segitiga siku-siku dengan sisi $a=3$ dan $b=4$. Hitung sisi miring $c$.

Rumus: $c = \sqrt{a^2 + b^2}$

````python
# Tulis kode Anda
import math

a, b = 3, 4
c = math.sqrt(a**2 + b**2)
print(f"Sisi miring = {c}")
````

### Soal 5: Luas Lingkaran

Hitung luas lingkaran dengan jari-jari $r = 7$.

Rumus: $L = \pi r^2$

````python
# Tulis kode Anda
import math

r = 7
luas = math.pi * r ** 2
print(f"Luas lingkaran = {luas:.2f}")
````

### Soal 6: Keliling Lingkaran

Hitung keliling lingkaran dengan diameter $d = 14$.

Rumus: $K = \pi \times d$

````python
# Tulis kode Anda
import math

d = 14
keliling = math.pi * d
print(f"Keliling lingkaran = {keliling:.2f}")
````

### Soal 7: Tinggi Tangga

Tangga sepanjang $10\ m$ bersandar di dinding dengan sudut $30\degree$.
Berapa tinggi dinding yang dicapai?

Rumus: $\text{tinggi} = \text{panjang} \times \sin(\theta)$

````python
# Tulis kode Anda
import math

panjang = 10
sudut = 30
tinggi = panjang * math.sin(math.radians(sudut))
print(f"Tinggi dinding = {tinggi:.1f} meter")
````

### Soal 8: Komponen Gaya

Gaya $F = 100\ N$ membentuk sudut $60\degree$ terhadap horizontal.
Hitung $F_x$ (horizontal) dan $F_y$ (vertikal).

$F_x = F \cdot \cos(\theta)$

$F_y = F \cdot \sin(\theta)$

````python
# Tulis kode Anda
import math

F = 100
sudut = 60
rad = math.radians(sudut)
Fx = F * math.cos(rad)
Fy = F * math.sin(rad)
print(f"Fx = {Fx:.1f} N")
print(f"Fy = {Fy:.1f} N")
````

### Soal 9: Cek Bilangan Genap/Ganjil

Buat program yang menerima sebuah angka, lalu mencetak apakah angka tersebut **genap** atau **ganjil**.

Petunjuk: Gunakan `if-else` dan operator modulo `%`.
- Jika `angka % 2 == 0` maka genap
- Selain itu ganjil

````python
# Tulis kode Anda
angka = int(input("Masukkan angka: "))

if angka % 2 == 0:
    print(f"{angka} adalah bilangan genap")
else:
    print(f"{angka} adalah bilangan ganjil")
````

### Soal 10: Deret Bilangan

Gunakan perulangan `for` untuk mencetak deret bilangan kelipatan 3 dari 3 sampai 30.

Output yang diharapkan:
```
3 6 9 12 15 18 21 24 27 30
```

````python
# Tulis kode Anda
# Gunakan for dan range(start, stop, step)
for i in range(3, 31, 3):
    print(i, end=" ")
````

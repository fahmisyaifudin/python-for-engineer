---
## Soal 1 — Luas Persegi

Luas persegi = sisi × sisi. Di Python, pangkat ditulis dengan operator `**`.

**s** = 5 cm → L = 5² = **25 cm²**

````python
# ── Jawaban Soal 1 ──
sisi = 5
luas = sisi ** 2

print(f"Luas persegi = {luas} cm²")
````

---
## Soal 2 — Volume Kubus

Sama seperti soal 1, tapi dipangkatkan **3** (rusuk × rusuk × rusuk).

**r** = 4 cm → V = 4³ = **64 cm³**

````python
# ── Jawaban Soal 2 ──
rusuk = 4
volume = rusuk ** 3

print(f"Volume kubus = {volume} cm³")
````

> 💡 Perhatikan bedanya dengan soal 1: `sisi ** 2` untuk persegi (2 dimensi),
> `rusuk ** 3` untuk kubus (3 dimensi).

---
## Soal 3 — Akar Kuadrat

`math.sqrt(x)` mengembalikan akar kuadrat dalam bentuk **float**, jadi hasilnya
`12.0` bukan `12`.

- √144 = **12.0**
- √225 = **15.0**

````python
# ── Jawaban Soal 3 ──
print(f"Akar dari 144 = {math.sqrt(144)}")
print(f"Akar dari 225 = {math.sqrt(225)}")
````

> 💡 Kalau ingin hasilnya bilangan bulat, bungkus dengan `int()`:
> `int(math.sqrt(144))` → `12`.

---
## Soal 4 — Pythagoras

Rumus: $c = \sqrt{a^2 + b^2}$

a = 3, b = 4 → c = √(3² + 4²) = √(9 + 16) = √25 = **5.0**

> 💡 Ini adalah **segitiga 3-4-5**, segitiga siku-siku istimewa yang paling sering
> muncul di soal. Sisi miringnya selalu bilangan bulat.

````python
# ── Jawaban Soal 4 ──
a, b = 3, 4
c = math.sqrt(a**2 + b**2)

print(f"Sisi miring = {c}")
````

---
## Soal 5 — Luas Lingkaran

Rumus: $L = \pi r^2$

r = 7 → L = π × 7² = π × 49 = **153.94**

`math.pi` sudah menyediakan nilai π dengan presisi tinggi — **jangan** ditulis manual
`3.14`, karena hasilnya jadi kurang teliti.

````python
# ── Jawaban Soal 5 ──
r = 7
luas = math.pi * r ** 2

print(f"Luas lingkaran = {luas:.2f}")
````

---
## Soal 6 — Keliling Lingkaran

Rumus: $K = \pi \times d$

d = 14 → K = π × 14 = **43.98**

> 💡 Diameter 14 = 2 × jari-jari 7. Jadi lingkaran di soal 5 dan soal 6 sebenarnya
> **lingkaran yang sama** — hanya yang satu ditanya luasnya, yang satu kelilingnya.
> Cek: K = 2πr = 2 × π × 7 = 43.98 ✓ (hasilnya sama)

````python
# ── Jawaban Soal 6 ──
d = 14
keliling = math.pi * d

print(f"Keliling lingkaran = {keliling:.2f}")
````

---
## Soal 7 — Tinggi Tangga

Rumus: $\text{tinggi} = \text{panjang} \times \sin(\theta)$

**Penting:** fungsi trigonometri di Python (`math.sin`, `math.cos`, `math.tan`)
menerima sudut dalam **radian**, bukan derajat. Karena itu sudutnya dikonversi dulu
dengan `math.radians()`.

panjang = 10 m, θ = 30° → tinggi = 10 × sin(30°) = 10 × 0.5 = **5.0 m**

````python
# ── Jawaban Soal 7 ──
panjang = 10
sudut = 30

tinggi = panjang * math.sin(math.radians(sudut))

print(f"Tinggi dinding = {tinggi:.1f} meter")
````

> 💡 Kalau lupa `math.radians()`, `math.sin(30)` akan dihitung sebagai sin(30 radian)
> ≈ -0.988 dan hasilnya jadi **negatif** — itu tanda Anda lupa konversi.

---
## Soal 8 — Komponen Gaya

F = 100 N, θ = 60°

- Fx = F cos θ = 100 × cos(60°) = 100 × 0.5 = **50.0 N**
- Fy = F sin θ = 100 × sin(60°) = 100 × 0.866 = **86.6 N**

> 💡 Cek cepat dengan Pythagoras: √(50² + 86.6²) ≈ 100 N — balik lagi ke besar
> gaya aslinya ✓

````python
# ── Jawaban Soal 8 ──
F = 100
sudut = 60
rad = math.radians(sudut)

Fx = F * math.cos(rad)
Fy = F * math.sin(rad)

print(f"Fx = {Fx:.1f} N")
print(f"Fy = {Fy:.1f} N")
````

````python
# ── Cek: Fx dan Fy harus menghasilkan F lagi (Pythagoras) ──
F_balik = math.sqrt(Fx**2 + Fy**2)

print(f"√(Fx² + Fy²) = {F_balik:.1f} N  → sama dengan F ✓")
````

## Soal 9 — Cek Bilangan Genap/Ganjil

Operator `%` (modulo) menghasilkan **sisa bagi**. Kalau `angka % 2` sisanya `0`,
berarti angka itu habis dibagi 2 → **genap**.

Soal ini memakai `input()`, jadi angkanya diketik saat program dijalankan. Supaya sel
di bawah bisa dijalankan otomatis (tanpa mengetik), nilai contohnya ditulis langsung
di kode — versi `input()`-nya tetap ada di komentar.

````python
# ── Jawaban Soal 9 ──
# Versi interaktif — pakai ini kalau dijalankan sendiri:
# angka = int(input("Masukkan angka: "))

# Versi nilai contoh, supaya sel ini bisa dijalankan otomatis:
angka = 7

if angka % 2 == 0:
    print(f"{angka} adalah bilangan genap")
else:
    print(f"{angka} adalah bilangan ganjil")
````

````python
# ── Cek untuk beberapa angka sekaligus ──
for angka in [7, 10, 0, 33, 100]:
    jenis = "genap" if angka % 2 == 0 else "ganjil"
    print(f"{angka:>4} → {jenis}")
````

> 💡 `0` juga **genap** — karena 0 habis dibagi 2 (sisanya 0). Ini sering bikin
> bingung, tapi aturannya tetap sama: cek sisanya, bukan besar angkanya.

## Soal 10 — Deret Bilangan

`range(3, 31, 3)` menghasilkan 3, 6, 9, 12, ... 30.

Perhatikan **stop-nya 31, bukan 30** — angka terakhir di `range()` tidak ikut.

Output yang diharapkan:
```
3 6 9 12 15 18 21 24 27 30
```

````python
# ── Jawaban Soal 10 ──
for i in range(3, 31, 3):
    print(i, end=" ")

print()  # pindah baris di akhir
````

````python
# ── Cara lain: kumpulkan dulu jadi satu teks, baru cetak sekali ──
deret = " ".join(str(i) for i in range(3, 31, 3))

print(deret)
````

> 💡 `print(i, end=" ")` membuat output tetap di baris yang sama, dipisah spasi.
> Tanpa `end=" "`, setiap angka akan turun ke baris baru.

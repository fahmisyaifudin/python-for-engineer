# 🧮 Operasi Vektor dengan NumPy

Materi: **Penjumlahan vektor**, **Perkalian skalar**, **Dot product**.

```python
import numpy as np
print(f'NumPy {np.__version__}')
```

---
## 🎯 Membuat Vektor

```python
# Vektor 2 dimensi
a = np.array([3, 4])
b = np.array([1, 2])
print(f'a = {a}, shape: {a.shape}')
print(f'b = {b}, shape: {b.shape}')
```

```python
# Vektor 3 dimensi
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
print(f'v1 = {v1}')
print(f'v2 = {v2}')
```

---
## 1⃣ Penjumlahan Vektor

> **a + b** = (a₁ + b₁, a₂ + b₂, ..., aₙ + bₙ)

```python
a = np.array([3, 4])
b = np.array([1, 2])
print(f'{a} + {b} = {a + b}')
```

```python
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
v3 = np.array([-1, 0, 2])
print(f'{v1} + {v2} + {v3} = {v1 + v2 + v3}')
```

```python
# Aplikasi: resultan gaya
F1 = np.array([10, 5])    # N
F2 = np.array([-3, 8])    # N
F_resultan = F1 + F2
print(f'F_resultan = {F_resultan} N')
```

---
## 2⃣ Perkalian Skalar

> **c · a** = (c·a₁, c·a₂, ..., c·aₙ)

```python
a = np.array([2, 3])
print(f'{a} x 3   = {a * 3}')
print(f'{a} x 0.5 = {a * 0.5}')
print(f'{a} x -1  = {a * -1}')
```

```python
# Aplikasi: F = m x a
m = 2.0                       # skalar
a = np.array([3.0, 4.0])      # vektor percepatan
F = m * a
print(f'F = {F} N')
```

---
## 3⃣ Dot Product

> **a · b** = Σ aᵢ·bᵢ

```python
a = np.array([2, 3])
b = np.array([4, 5])
print(f'np.dot({a}, {b}) = {np.dot(a, b)}')
print(f'{a} @ {b} = {a @ b}')
```

```python
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
print(f'np.dot({v1}, {v2}) = {np.dot(v1, v2)}')
```

```python
# Dot = 0 -> tegak lurus
a = np.array([1, 0])
b = np.array([0, 1])
print(f'{a} . {b} = {np.dot(a, b)} -> orthogonal')

c = np.array([2, 4])
d = np.array([-2, 1])
print(f'{c} . {d} = {np.dot(c, d)} -> orthogonal')
```

```python
# Aplikasi: W = F . s
F = np.array([10, 5])     # N
s = np.array([4, 2])      # m
W = np.dot(F, s)
print(f'W = {W} Joule')
```

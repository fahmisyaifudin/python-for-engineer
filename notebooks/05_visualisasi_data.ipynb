# 📈 Visualisasi Data dengan Matplotlib

Matplotlib adalah pustaka visualisasi data paling populer di Python. Notebook ini mencakup:

1. **Dasar Matplotlib** — setup, figure, dan axes
2. **Line Plot** — grafik garis untuk data kontinu
3. **Bar Chart** — diagram batang untuk data kategorikal
4. **Histogram** — distribusi data numerik
5. **Scatter Plot** — hubungan antar dua variabel
6. **Pie Chart** — proporsi / persentase

---
## 0️⃣ Setup & Import

Pastikan matplotlib sudah terinstall. Jika belum: `pip install matplotlib`

```python
# ── Import pustaka ──
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Agar plot tampil di notebook
%matplotlib inline

print(f"Matplotlib version: {plt.matplotlib.__version__}")
```

---
## 1️⃣ Dasar Matplotlib

Setiap plot di matplotlib terdiri dari:
- **Figure** — kanvas/wadah utama
- **Axes** — area plot sebenarnya (satu figure bisa punya beberapa axes)

```python
# ── Plot pertama: figure eksplisit ──

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

fig, ax = plt.subplots()     # buat figure + axes
ax.plot(x, y)                 # plot garis
ax.set_title("Plot Pertama Saya")
ax.set_xlabel("Sumbu X")
ax.set_ylabel("Sumbu Y")
plt.show()
```

```python
# ── Figure size & style ──

fig, ax = plt.subplots(figsize=(8, 5))  # ukuran figure: lebar 8, tinggi 5

x = np.linspace(0, 10, 100)
y = np.sin(x)

ax.plot(x, y, color="blue", linewidth=2, linestyle="-", label="sin(x)")
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title("Gelombang Sinus", fontsize=14)
ax.set_xlabel("x", fontsize=12)
ax.set_ylabel("sin(x)", fontsize=12)
plt.show()
```

---
## 2️⃣ Line Plot

Line plot cocok untuk data **kontinu** atau **time series**.

```python
# ── Line plot: multiple lines ──

x = np.linspace(0, 10, 200)

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(x, np.sin(x), label="sin(x)", linewidth=2)
ax.plot(x, np.cos(x), label="cos(x)", linewidth=2, linestyle="--")
ax.plot(x, np.sin(x) + np.cos(x), label="sin(x)+cos(x)", linewidth=1.5, linestyle=":")

ax.legend(fontsize=12)
ax.grid(True, alpha=0.3)
ax.set_title("Fungsi Trigonometri", fontsize=14)
ax.set_xlabel("x", fontsize=12)
ax.set_ylabel("y", fontsize=12)
ax.set_xlim(0, 10)
ax.set_ylim(-2, 2)
plt.show()
```

```python
# ── Line plot dengan marker ──

x = [1, 2, 3, 4, 5, 6, 7]
y = [23, 45, 56, 78, 65, 54, 43]

fig, ax = plt.subplots(figsize=(8, 4))

# Berbagai gaya marker dan line
ax.plot(x, y, marker="o", markersize=8, linewidth=2,
        color="darkblue", markerfacecolor="red",
        markeredgecolor="darkred", markeredgewidth=2)

# Anotasi nilai
for i, (xi, yi) in enumerate(zip(x, y)):
    ax.annotate(f"{yi}", (xi, yi), textcoords="offset points",
                xytext=(0, 10), ha="center", fontsize=9)

ax.set_title("Data Penjualan Mingguan", fontsize=14)
ax.set_xlabel("Hari", fontsize=12)
ax.set_ylabel("Penjualan (unit)", fontsize=12)
ax.grid(True, alpha=0.3)
plt.show()
```

---
## 3️⃣ Bar Chart

Bar chart cocok untuk membandingkan data **kategorikal**.

```python
# ── Bar chart vertikal ──

kategori = ["IT", "HR", "Finance", "Marketing", "Operasional"]
nilai = [85, 72, 90, 78, 88]

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(kategori, nilai, color=["#3498db", "#e74c3c", "#2ecc71", "#f39c12", "#9b59b6"])

# Tambahkan nilai di atas bar
for bar in bars:
    height = bar.get_height()
    ax.annotate(f"{height}",
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 5), textcoords="offset points",
                ha="center", fontsize=11)

ax.set_title("Rata-rata Kepuasan per Departemen", fontsize=14)
ax.set_xlabel("Departemen", fontsize=12)
ax.set_ylabel("Skor Kepuasan", fontsize=12)
ax.set_ylim(0, 100)
ax.grid(axis="y", alpha=0.3)
plt.show()
```

---
## 4️⃣ Histogram

Histogram menunjukkan **distribusi** data numerik — seberapa sering nilai muncul dalam rentang tertentu.

```python
# ── Histogram dasar ──

np.random.seed(42)
data = np.random.randn(1000)  # 1000 sampel distribusi normal

fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(data, bins=30, color="#3498db", edgecolor="white", alpha=0.7)

ax.set_title("Distribusi Normal (1000 sampel)", fontsize=14)
ax.set_xlabel("Nilai", fontsize=12)
ax.set_ylabel("Frekuensi", fontsize=12)
ax.grid(axis="y", alpha=0.3)
plt.show()
```

---
## 5️⃣ Scatter Plot

Scatter plot menunjukkan **hubungan** antara dua variabel numerik.

```python
# ── Scatter plot dasar ──

np.random.seed(42)
x = np.random.rand(50) * 10          # 0 - 10
y = 2 * x + np.random.randn(50) * 2  # y = 2x + noise

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(x, y, color="#3498db", alpha=0.7, s=80, edgecolor="navy", linewidth=0.5)

ax.set_title("Hubungan X dan Y", fontsize=14)
ax.set_xlabel("X", fontsize=12)
ax.set_ylabel("Y", fontsize=12)
ax.grid(True, alpha=0.3)
plt.show()
```

---
## 6️⃣ Pie Chart

Pie chart menunjukkan **proporsi** atau persentase dari keseluruhan.

```python
# ── Pie chart ──

label = ["IT", "Marketing", "Finance", "HR", "Operasional"]
ukuran = [35, 25, 20, 10, 10]  # persentase
warna = ["#3498db", "#e74c3c", "#2ecc71", "#f39c12", "#9b59b6"]

fig, ax = plt.subplots(figsize=(8, 6))

ax.pie(ukuran, labels=label, colors=warna, autopct="%1.1f%%",
       startangle=90, explode=(0.05, 0, 0, 0, 0),
       shadow=True, textprops={"fontsize": 12})

ax.set_title("Komposisi Karyawan per Departemen", fontsize=14)
plt.show()
```

---
## 📚 Rangkuman

| Jenis Plot | Kegunaan | Fungsi |
|-----------|----------|--------|
| **Line Plot** | Data kontinu / time series | `ax.plot()` |
| **Bar Chart** | Membandingkan kategori | `ax.bar()`, `ax.barh()` |
| **Histogram** | Distribusi data numerik | `ax.hist()` |
| **Scatter Plot** | Hubungan dua variabel | `ax.scatter()` |
| **Pie Chart** | Proporsi / persentase | `ax.pie()` |

### Poin Penting
- **Figure** adalah kanvas, **Axes** adalah area plot
- Gunakan `figsize` untuk mengontrol ukuran plot
- `label` + `legend()` untuk legenda, `grid()` untuk grid
- `plt.tight_layout()` mencegah elemen saling bertumpuk
- `plt.style.use()` untuk mengganti tema secara global
- Ekspor plot dengan `fig.savefig("file.png", dpi=300)`

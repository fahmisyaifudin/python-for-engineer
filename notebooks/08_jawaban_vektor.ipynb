## Soal 1 — Penjumlahan Vektor

**p** = (7, -2, 4), **q** = (-3, 5, 1)

Penjumlahan vektor dikerjakan **elemen per elemen** — cukup pakai operator `+`.

Secara manual: (7 + (-3), -2 + 5, 4 + 1)

```python
# ── Jawaban Soal 1 ──
p = np.array([7, -2, 4])
q = np.array([-3, 5, 1])

hasil = p + q

print(f"p = {p}")
print(f"q = {q}")
print(f"p + q = {hasil}")
print(f"\nCek manual: (7 + (-3), -2 + 5, 4 + 1) = (4, 3, 5)")
```

## Soal 2 — Perkalian Skalar

**v** = (2, -1, 3, 0)

Kalikan **setiap elemen** dengan skalar tersebut.

```python
# ── Jawaban Soal 2 ──
v = np.array([2, -1, 3, 0])

print(f"v     = {v}")
print(f"3v    = {3 * v}")
print(f"-2v   = {-2 * v}")
print(f"0.5v  = {0.5 * v}")
```

Hasilnya:

| Operasi | Hasil |
|---|---|
| **3v** | (6, -3, 9, 0) |
| **-2v** | (-4, 2, -6, 0) |
| **0.5v** | (1, -0.5, 1.5, 0) |

> 💡 Perhatikan **-2v** arahnya berbalik 180° dari **v**, dan **0.5v** panjangnya jadi separuh.
> Perkalian skalar mengubah **panjang** (dan mungkin arah), bukan dimensi.

## Soal 3 — Dot Product

**a** = (4, -1, 2), **b** = (0, 3, 5)

Dot product = jumlah hasil kali elemen yang bersesuaian, dan hasilnya **skalar** (bukan vektor).

Secara manual: (4×0) + (-1×3) + (2×5) = 0 - 3 + 10 = **7**

```python
# ── Jawaban Soal 3 ──
a = np.array([4, -1, 2])
b = np.array([0, 3, 5])

print(f"np.dot(a, b) = {np.dot(a, b)}")
print(f"a @ b        = {a @ b}")
print(f"\nKeduanya sama — operator @ adalah cara singkat untuk np.dot() pada vektor 1D.")
```

## Soal 4 — Cek Orthogonal

Dua vektor **tegak lurus** kalau dot product-nya **= 0**.

1. **u** = (2, -3), **v** = (3, 2) → (2×3) + (-3×2) = 6 - 6 = **0**
2. **p** = (1, 4, -2), **q** = (3, -1, 2) → (1×3) + (4×-1) + (-2×2) = 3 - 4 - 4 = **-5**

```python
# ── Jawaban Soal 4 ──
u = np.array([2, -3])
v = np.array([3, 2])

p = np.array([1, 4, -2])
q = np.array([3, -1, 2])

for nama, x, y in [("u . v", u, v), ("p . q", p, q)]:
    dot = np.dot(x, y)
    status = "ORTHOGONAL (tegak lurus) ✓" if dot == 0 else "TIDAK orthogonal ✗"
    print(f"{nama} = {dot:>3}  → {status}")
```

**Kesimpulan:**

1. **u** dan **v** → **tegak lurus** (dot = 0)
2. **p** dan **q** → **tidak tegak lurus** (dot = -5, bukan 0)

> 💡 Dot negatif artinya sudut antara keduanya **lebih dari 90°** (agak berlawanan arah).

## Soal 5 — Proyeksi Vektor

Rumusnya:

> proj = ((a · b) / (b · b)) × b

**a** = (6, 2), **b** = (3, 4)

Hitung dulu bagian skalarnya:
- a · b = (6×3) + (2×4) = 18 + 8 = **26**
- b · b = (3×3) + (4×4) = 9 + 16 = **25**
- skalar = 26 / 25 = **1.04**

Lalu kalikan ke **b**: 1.04 × (3, 4) = **(3.12, 4.16)**

```python
# ── Jawaban Soal 5 ──
a = np.array([6, 2])
b = np.array([3, 4])

# Rumus proyeksi: ((a . b) / (b . b)) * b
skalar = np.dot(a, b) / np.dot(b, b)
proyeksi = skalar * b

print(f"a . b        = {np.dot(a, b)}")
print(f"b . b        = {np.dot(b, b)}")
print(f"skalar       = {np.dot(a, b)} / {np.dot(b, b)} = {skalar}")
print(f"\nproyeksi a pada b = {skalar} x {b} = {proyeksi}")
```

```python
# ── Cek: hasil proyeksi harus sejajar dengan b ──
# Kalau sejajar, hasil bagi tiap elemennya sama.
print(f"proyeksi / b = {proyeksi / b}")
print(f"\nKeduanya {proyeksi[0] / b[0]} → proyeksi memang searah dengan b ✓")
```

> 💡 **Apa artinya?** Proyeksi adalah "bayangan" vektor **a** kalau disorot tegak lurus ke arah
> **b**. Panjang bayangannya = 1.04 × panjang b.

## Soal 6 — Cosine Similarity

Rumusnya:

> cos(θ) = (a · b) / (||a|| × ||b||)

- **x** = (3, 1, 4, 1, 5)
- **y** = (1, 5, 1, 3, 2)

`np.linalg.norm()` menghitung **panjang** vektor (norma Euclidean): √(Σ elemen²).

Nilai cos selalu antara **-1 dan 1**:
- **1** → arahnya sama persis
- **0** → tegak lurus, tidak berhubungan
- **-1** → arahnya berlawanan

```python
# ── Jawaban Soal 6 ──
x = np.array([3, 1, 4, 1, 5])
y = np.array([1, 5, 1, 3, 2])

dot = np.dot(x, y)
norm_x = np.linalg.norm(x)
norm_y = np.linalg.norm(y)

cos_sim = dot / (norm_x * norm_y)

print(f"x . y            = {dot}")
print(f"||x||            = {norm_x:.4f}")
print(f"||y||            = {norm_y:.4f}")
print(f"\ncosine similarity = {dot} / ({norm_x:.4f} x {norm_y:.4f})")
print(f"                  = {cos_sim:.4f}")
```

```python
# ── Ubah jadi sudut (derajat) supaya lebih mudah dibaca ──
sudut = np.degrees(np.arccos(cos_sim))

print(f"cos(θ) = {cos_sim:.4f}")
print(f"θ      = {sudut:.1f}°")
```

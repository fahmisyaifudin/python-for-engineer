## 1️⃣ Series & DataFrame

Dua struktur data utama di pandas:
- **Series** — satu dimensi (seperti list/array dengan label)
- **DataFrame** — dua dimensi (seperti tabel/spreadsheet)

```python
# ── Series ──
# Series mirip list tapi memiliki index label

nilai = pd.Series([85, 92, 78, 90, 88])
print("Series sederhana:")
print(nilai)
print()

# Series dengan index kustom
nilai_label = pd.Series(
    [85, 92, 78, 90, 88],
    index=["Budi", "Rina", "Charlie", "Dewi", "Eko"]
)
print("Series dengan index:")
print(nilai_label)
print()

# Akses elemen
print(f"Nilai Budi : {nilai_label['Budi']}")
print(f"Nilai Rina : {nilai_label['Rina']}")
```

```python
# ── DataFrame dari dictionary ──
# Cara paling umum: dictionary dengan key = kolom

data_mahasiswa = {
    "nama"   : ["Budi", "Rina", "Charlie", "Dewi", "Eko"],
    "jurusan": ["Teknik Mesin", "Teknik Elektro", "Teknik Sipil",
                 "Teknik Mesin", "Teknik Elektro"],
    "nilai"  : [85, 92, 78, 90, 88],
    "semester": [4, 6, 4, 6, 4],
}

df = pd.DataFrame(data_mahasiswa)
print("DataFrame Mahasiswa:")
df
```

```python
# ── Properti dasar DataFrame ──

print(f"Shape (baris, kolom) : {df.shape}")
print(f"Jumlah baris         : {len(df)}")
print(f"Jumlah kolom         : {len(df.columns)}")
print(f"Nama kolom           : {list(df.columns)}")
print(f"Tipe data per kolom  :")
print(df.dtypes)
```

## 2️⃣ Membaca Data dari File

Pandas dapat membaca berbagai format file. Kita akan membuat contoh data CSV terlebih dahulu.

```python
# ── Membuat contoh file CSV ──

contoh_csv = """nama,usia,kota,gaji
Budi,28,Jakarta,8500000
Rina,25,Bandung,7200000
Charlie,32,Surabaya,9500000
Dewi,27,Jakarta,8000000
Eko,30,Bandung,7800000
Fajar,35,Surabaya,11000000
Gita,29,Jakarta,9000000
Hadi,31,Bandung,8200000
Intan,26,Surabaya,7500000
Joko,33,Jakarta,10500000
"""

with open("data/data_karyawan.csv", "w") as f:
    f.write(contoh_csv)

print("File data_karyawan.csv berhasil dibuat ✓")
```

## 3️⃣ Eksplorasi Data

Setelah data dimuat, langkah pertama adalah **memahami strukturnya**.

```python
# ── Melihat data: head() & tail() ──

print("── 5 baris pertama (head) ──")
print(df_karyawan.head())

print("\n── 3 baris pertama (head(3)) ──")
print(df_karyawan.head(3))

print("\n── 3 baris terakhir (tail(3)) ──")
print(df_karyawan.tail(3))
```

```python
# ── Info & Statistik Deskriptif ──

print("── info() — tipe data & non-null count ──")
print(df_karyawan.info())

print("\n── describe() — statistik numerik ──")
print(df_karyawan.describe())
```

```python
# ── Unique, value_counts, dan ringkasan kategorikal ──

print("Kota yang ada:", df_karyawan["kota"].unique())
print("Jumlah tiap kota:")
print(df_karyawan["kota"].value_counts())
print()

print("Statistik kolom kategorikal:")
print(df_karyawan.describe(include=["object"]))
```

---
## 4️⃣ Seleksi & Filtering

Memilih baris/kolom tertentu dari DataFrame.

```python
# ── Memilih kolom ──

# Satu kolom → Series
print("── Kolom 'nama' (Series) ──")
print(df_karyawan["nama"])

print("\n── Beberapa kolom (DataFrame) ──")
print(df_karyawan[["nama", "gaji"]])
```

```python
# ── Seleksi baris: iloc (index) vs loc (label) ──

print("── iloc[0] — baris pertama (index 0) ──")
print(df_karyawan.iloc[0])

print("\n── iloc[1:4] — baris 1 s/d 3 ──")
print(df_karyawan.iloc[1:4])

print("\n── iloc[0, 2] — baris 0, kolom 2 ──")
print(df_karyawan.iloc[0, 2])

print("\n── iloc[:, 0:3] — semua baris, kolom 0-2 ──")
print(df_karyawan.iloc[:, 0:3])
```

```python
# ── Boolean indexing (filtering dengan kondisi) ──

# Karyawan dengan gaji > 8 juta
print("── Gaji > 8.000.000 ──")
print(df_karyawan[df_karyawan["gaji"] > 8000000])

print("\n── Karyawan di Jakarta ──")
print(df_karyawan[df_karyawan["kota"] == "Jakarta"])

print("\n── Karyawan di Jakarta dengan gaji > 8jt ──")
kondisi = (df_karyawan["kota"] == "Jakarta") & (df_karyawan["gaji"] > 8000000)
print(df_karyawan[kondisi])

print("\n── Karyawan di Bandung atau Surabaya ──")
print(df_karyawan[df_karyawan["kota"].isin(["Bandung", "Surabaya"])])
```

```python
# ── Method query() — alternatif yang lebih bersih ──

print("── query: gaji > 8jt ──")
print(df_karyawan.query("gaji > 8000000"))

print("\n── query: Jakarta dan gaji > 8jt ──")
print(df_karyawan.query("kota == 'Jakarta' and gaji > 8000000"))
```

## 📚 Rangkuman

| Konsep | Method / Cara | Contoh |
|--------|--------------|--------|
| **Membaca CSV** | `pd.read_csv()` | `df = pd.read_csv("file.csv")` |
| **Melihat data** | `head()`, `tail()`, `info()`, `describe()` | `df.head(10)` |
| **Seleksi kolom** | `df["kolom"]`, `df[["a","b"]]` | `df["nama"]` |
| **Seleksi baris** | `iloc[]`, `loc[]`, `query()` | `df.iloc[0:5]` |
| **Filtering** | Boolean indexing | `df[df["gaji"] > 10000]` |

## 1️⃣ Variabel & Tipe Data

Variabel menyimpan data sehingga bisa digunakan dan dimanipulasi. Python **dynamically typed** — Anda tidak perlu mendeklarasikan tipe secara eksplisit.

```python
# ── Angka (Number) ──
usia = 25               # int — bilangan bulat
phi = 3.14159           # float — bilangan desimal
bil_kompleks = 1 + 2j   # complex — digunakan di engineering

print(f"usia       = {usia}  ({type(usia).__name__})")
print(f"phi        = {phi}  ({type(phi).__name__})")
print(f"kompleks   = {bil_kompleks}  ({type(bil_kompleks).__name__})")
```

```python
# ── Teks (String) ──
nama = "Budi"
salam = f"Halo, {nama}!"   # f-string — sisipkan variabel ke dalam teks
teks_multi = """Ini adalah
teks multi-baris."""

print(salam)
print("Huruf kapital:", nama.upper())
print("Panjang:", len(nama))
print(teks_multi)
```

```python
# ── Boolean ──
siap = True
selesai = False

print(f"Siap: {siap}, Selesai: {selesai}")
print(f"TIDAK siap: {not siap}")
print(f"Siap DAN Selesai: {siap and selesai}")
print(f"Siap ATAU Selesai: {siap or selesai}")
```

```python
# ── List (terurut, bisa diubah) ──
buah = ["apel", "pisang", "ceri"]
buah.append("kurma")          # tambah item
buah[0] = "alpukat"           # ubah item

print("Buah:", buah)
print("Pertama:", buah[0])
print("Terakhir:", buah[-1])
print("Irisan [1:3]:", buah[1:3])
```

```python
# ── Dictionary (pasangan kunci-nilai) ──
mahasiswa = {
    "nama": "Rina",
    "usia": 22,
    "matkul": ["Matematika", "Fisika"],
}

print("Nama:", mahasiswa["nama"])
print("Usia:", mahasiswa.get("usia", "tidak diketahui"))
mahasiswa["nilai"] = "A"         # tambah kunci baru
print("Setelah ditambah:", mahasiswa)
```

```python
# ── Tuple (terurut, tidak bisa diubah) & Set (tidak urut, unik) ──
koordinat = (10.0, 20.5)   # tuple — tidak bisa diubah setelah dibuat
angka_unik = {1, 2, 2, 3, 3, 3}  # set — duplikat otomatis dihapus

print("Koordinat:", koordinat)
print("Angka unik:", angka_unik)
```

### ✏️ Latihan — Variabel

Buat variabel `radius = 5` dan hitung luas lingkaran ($\pi r^2$).

```python
# Tulis kode Anda di sini
import math

radius = 5
luas = math.pi * radius ** 2
print(f"Luas lingkaran dengan r={radius} adalah {luas:.2f}")
```

## 2️⃣ Percabangan (if-elif-else)

Kendalikan alur program dengan memeriksa kondisi.

```python
# ── If-else dasar ──
suhu = 30

if suhu > 35:
    print("🔥 Sangat panas!")
elif suhu > 25:
    print("☀️ Hari yang hangat")
elif suhu > 15:
    print("🌤️ Cuaca nyaman")
else:
    print("❄️ Hari yang sejuk")

print("---")
# ── Ternary (if-else satu baris) ──
status = "dewasa" if usia >= 18 else "anak-anak"
print(f"Status: {status}")
```

```python
# ── Menggabungkan kondisi ──
nilai = 85
kehadiran = 90

if nilai >= 80 and kehadiran >= 75:
    print("✅ Lulus dengan pujian")
elif nilai >= 60 or kehadiran >= 75:
    print("✅ Lulus bersyarat")
else:
    print("❌ Tidak lulus")
```

```python
# ── Pemeriksaan keanggotaan & identitas ──
item = [1, 2, 3, 4, 5]

if 3 in item:
    print("3 ada di dalam list")

a = [1, 2]
b = [1, 2]
print("a == b:", a == b)    # True — nilainya sama
print("a is b:", a is b)    # False — objek berbeda
print("a is not b:", a is not b)
```

## 3️⃣ Perulangan (Loop)

Ulangi tindakan tanpa menyalin kode berulang kali.

```python
# ── for loop — iterasi melalui urutan ──
print("── for loop pada list ──")
for buah in ["apel", "pisang", "ceri"]:
    print(f"  - {buah}")

print("\n── for loop dengan range ──")
for i in range(5):        # 0, 1, 2, 3, 4
    print(f"  i = {i}")

print("\n── range dengan start, stop, step ──")
for i in range(2, 10, 2):  # 2, 4, 6, 8
    print(f"  genap = {i}")
```

```python
# ── enumerate — dapatkan indeks dan nilai ──
warna = ["merah", "hijau", "biru"]

for idx, w in enumerate(warna):
    print(f"  {idx}: {w}")
```

```python
# ── Iterasi dictionary ──
nilai_siswa = {"Budi": 85, "Rina": 92, "Charlie": 78}

for nama, nilai in nilai_siswa.items():
    print(f"  {nama}: {nilai}")
```

## 4️⃣ Fungsi

Fungsi memungkinkan Anda mengemas logika yang dapat digunakan kembali, menghindari pengulangan, dan menjaga kode tetap rapi.

```python
# ── Mendefinisikan fungsi ──
def sapa(nama):
    """Mengembalikan string sapaan."""
    return f"Halo, {nama}!"

def sapa_lengkap(nama, sapaan="Hai"):
    """Fungsi dengan parameter default."""
    return f"{sapaan}, {nama}!"

print(sapa("Budi"))
print(sapa_lengkap("Rina"))
print(sapa_lengkap("Charlie", "Hey"))
```

```python
# ── Banyak parameter & nilai kembali ──
def persegi_panjang(p, l):
    """Mengembalikan luas dan keliling persegi panjang."""
    luas = p * l
    keliling = 2 * (p + l)
    return luas, keliling          # mengembalikan tuple

luas, kel = persegi_panjang(5, 3)
print(f"Luas: {luas}, Keliling: {kel}")
```

```python
# ── *args & **kwargs (argumen variabel) ──
def perkenalkan(**kwargs):
    """Menerima berapapun argumen kata kunci."""
    for kunci, nilai in kwargs.items():
        print(f"  {kunci}: {nilai}")

perkenalkan(nama="Budi", usia=25, kota="Jakarta")

def jumlahkan(*args):
    """Menerima berapapun argumen posisi."""
    return sum(args)

print(f"jumlahkan(1, 2, 3, 4, 5) = {jumlahkan(1, 2, 3, 4, 5)}")
```

```python
# ── Lambda (fungsi anonim) ──
kuadrat = lambda x: x ** 2
print(f"kuadrat(5) = {kuadrat(5)}")

# Penggunaan umum: sorting dengan kunci
titik = [(1, 5), (3, 2), (2, 8)]
titik_terurut = sorted(titik, key=lambda t: t[1])  # urut berdasarkan y
print(f"Diurutkan berdasarkan y: {titik_terurut}")
```

```python
# ── Docstring & type hints ──
def celsius_ke_fahrenheit(c: float) -> float:
    """Konversi Celsius ke Fahrenheit.

    Args:
        c: Suhu dalam Celsius.

    Returns:
        Suhu dalam Fahrenheit.
    """
    return (c * 9 / 5) + 32

print(f"0°C = {celsius_ke_fahrenheit(0)}°F")
print(f"100°C = {celsius_ke_fahrenheit(100)}°F")
print(f"37°C = {celsius_ke_fahrenheit(37)}°F")
```

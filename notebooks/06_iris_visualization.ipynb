# Visualisasi Dataset Iris

Dataset Iris (R. A. Fisher, 1936) adalah dataset klasik untuk machine learning dan visualisasi data. Berisi 150 sampel bunga iris dari 3 spesies:

| Spesies | Warna | Jumlah Sampel |
|---------|-------|---------------|
| **Setosa** | 🔵 Biru | 50 |
| **Versicolor** | 🔴 Merah | 50 |
| **Virginica** | 🟢 Hijau | 50 |

**Fitur (cm):**
- `sepal.length` — panjang sepal (kelopak)
- `sepal.width` — lebar sepal
- `petal.length` — panjang petal (mahkota)
- `petal.width` — lebar petal

> 💡 **Jalankan setiap sel** dengan memilihnya dan tekan `Shift + Enter`.

```python
# ── Import pustaka ──
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

%matplotlib inline

print("✅ Semua pustaka berhasil diimport")
```

```python
# ── Memuat dataset Iris ──
df = pd.read_csv("data/iris.csv")

# Bersihkan nama kolom (hapus spasi dan kapital)
df.columns = df.columns.str.strip()
df["variety"] = df["variety"].str.strip()

print(f"Dataset: {df.shape[0]} baris x {df.shape[1]} kolom")
print(f"Spesies: {df['variety'].unique()}")
print()
df.head()
```

---
## 1️⃣ Scatter Plot — Petal Length vs Petal Width

Scatter plot yang membandingkan panjang petal dan lebar petal.
- **Setosa** → 🔵 Biru
- **Versicolor** → 🔴 Merah
- **Virginica** → 🟢 Hijau

```python
# ── Scatter plot: Petal Length vs Petal Width ──

# Pisahkan data per spesies
setosa = df[df["variety"] == "Setosa"]
versicolor = df[df["variety"] == "Versicolor"]
virginica = df[df["variety"] == "Virginica"]

fig, ax = plt.subplots(figsize=(10, 7))

# Plot masing-masing spesies dengan warna berbeda
ax.scatter(setosa["petal.length"], setosa["petal.width"],
           color="blue", label="Setosa", alpha=0.8, s=60, edgecolor="black", linewidth=0.5)
ax.scatter(versicolor["petal.length"], versicolor["petal.width"],
           color="red", label="Versicolor", alpha=0.8, s=60, edgecolor="black", linewidth=0.5)
ax.scatter(virginica["petal.length"], virginica["petal.width"],
           color="green", label="Virginica", alpha=0.8, s=60, edgecolor="black", linewidth=0.5)

ax.set_title("Iris Dataset — Petal Length vs Petal Width", fontsize=14, fontweight="bold")
ax.set_xlabel("Panjang Petal (cm)", fontsize=12)
ax.set_ylabel("Lebar Petal (cm)", fontsize=12)
ax.legend(fontsize=11, loc="lower right")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

---
## 2️⃣ Scatter Plot — Sepal Length vs Sepal Width

Perbandingan dimensi sepal antar spesies.

```python
# ── Scatter plot: Sepal Length vs Sepal Width ──

fig, ax = plt.subplots(figsize=(10, 7))

ax.scatter(setosa["sepal.length"], setosa["sepal.width"],
           color="blue", label="Setosa", alpha=0.8, s=60, edgecolor="black", linewidth=0.5)
ax.scatter(versicolor["sepal.length"], versicolor["sepal.width"],
           color="red", label="Versicolor", alpha=0.8, s=60, edgecolor="black", linewidth=0.5)
ax.scatter(virginica["sepal.length"], virginica["sepal.width"],
           color="green", label="Virginica", alpha=0.8, s=60, edgecolor="black", linewidth=0.5)

ax.set_title("Iris Dataset — Sepal Length vs Sepal Width", fontsize=14, fontweight="bold")
ax.set_xlabel("Panjang Sepal (cm)", fontsize=12)
ax.set_ylabel("Lebar Sepal (cm)", fontsize=12)
ax.legend(fontsize=11, loc="upper right")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

---
## 3️⃣ Subplots — 4 Kombinasi Fitur

Semua kombinasi fitur dalam satu figure.

```python
# ── Subplots 2x2: semua kombinasi fitur ──

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

kombinasi = [
    ("petal.length", "petal.width", "Petal"),
    ("sepal.length", "sepal.width", "Sepal"),
    ("sepal.length", "petal.length", "Sepal Length vs Petal Length"),
    ("sepal.width", "petal.width", "Sepal Width vs Petal Width"),
]

for ax, (x_col, y_col, title) in zip(axes.flat, kombinasi):
    ax.scatter(setosa[x_col], setosa[y_col],
               color="blue", label="Setosa", alpha=0.8, s=50, edgecolor="black", linewidth=0.5)
    ax.scatter(versicolor[x_col], versicolor[y_col],
               color="red", label="Versicolor", alpha=0.8, s=50, edgecolor="black", linewidth=0.5)
    ax.scatter(virginica[x_col], virginica[y_col],
               color="green", label="Virginica", alpha=0.8, s=50, edgecolor="black", linewidth=0.5)
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.set_xlabel(f"{x_col} (cm)", fontsize=10)
    ax.set_ylabel(f"{y_col} (cm)", fontsize=10)
    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=9, loc="best")

fig.suptitle("Iris Dataset — Semua Kombinasi Fitur", fontsize=16, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

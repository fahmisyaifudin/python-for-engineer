## 1⃣ Kuadrat & Pangkat

Python memiliki beberapa cara untuk menghitung bilangan berpangkat.

```python
# --- Kuadrat & Pangkat ---
a = 5
kuadrat = a ** 2        # 5 pangkat 2 = 25
kubik   = a ** 3        # 5 pangkat 3 = 125
pangk4  = a ** 4        # 5 pangkat 4 = 625

print(f"{a}\u00b2  = {kuadrat}")
print(f"{a}\u00b3  = {kubik}")
print(f"{a}\u2074  = {pangk4}")

# --- Pangkat bilangan negatif ---
print(f"2\u207b\u00b2 = {2 ** -2}")   # 0.25
print(f"2\u207b\u00b3 = {2 ** -3}")   # 0.125
```

```python
# --- Fungsi pow() ---
print(f"pow(5, 2) = {pow(5, 2)}")   # 25
print(f"pow(5, 3) = {pow(5, 3)}")   # 125
print(f"pow(2, 5) = {pow(2, 5)}")   # 32

# pow() dengan modulus
print(f"pow(5, 2, 10) = {pow(5, 2, 10)}")  # (5^2) % 10 = 5
```

```python
# --- Fungsi matematika dari pustaka math ---
import math

# math.pow() -- selalu mengembalikan float
print(f"math.pow(5, 2) = {math.pow(5, 2)}")   # 25.0
print(f"math.pow(2, 3) = {math.pow(2, 3)}")   # 8.0

# Perbedaan: ** bisa untuk int, math.pow() selalu float
print(f"Tipe 5**2   : {type(5**2).__name__}")
print(f"Tipe pow(): {type(math.pow(5,2)).__name__}")
```

```python
# --- Aplikasi: Rumus Fisika ---
# Energi kinetik: E = 1/2 m v^2
massa = 10       # kg
kecepatan = 5     # m/s
ek = 0.5 * massa * kecepatan ** 2
print(f"Energi kinetik: {ek} joule")

# Hukum gravitasi Newton: F = G m1 m2 / r^2
G = 6.674e-11
m1, m2 = 100, 100
r = 10
F = G * m1 * m2 / r ** 2
print(f"Gaya gravitasi: {F:.4e} N")
```

## 2⃣ Akar (Square Root)

Akar kuadrat dan akar pangkat-n dapat dihitung dengan beberapa cara.

```python
# --- math.sqrt() --- akar kuadrat ---
import math

angka = 25
hasil = math.sqrt(angka)
print(f"Akar kuadrat dari {angka} = {hasil}")

# Beberapa contoh
for n in [4, 9, 16, 25, 100]:
    print(f"  akar({n:3d}) = {math.sqrt(n):.2f}")
```

```python
# --- Akar pangkat-n dengan ** ---
# Akar kuadrat = pangkat 1/2
print(f"Akar kuadrat 25 = {25 ** 0.5}")

# Akar kubik = pangkat 1/3
print(f"Akar kubik 27 = {27 ** (1/3):.4f}")
print(f"Akar kubik 64 = {64 ** (1/3):.4f}")

# Akar pangkat-n
print(f"Akar pangkat 4 dari 16 = {16 ** (1/4):.4f}")
print(f"Akar pangkat 5 dari 32 = {32 ** (1/5):.4f}")
```

```python
# --- Aplikasi: Teorema Pythagoras ---
import math

# c^2 = a^2 + b^2  ->  c = akar(a^2 + b^2)
a, b = 3, 4
c = math.sqrt(a**2 + b**2)
print(f"Segitiga {a}, {b}, {c} -> sisi miring = {c}")

# Jarak dua titik: d = akar((x2-x1)^2 + (y2-y1)^2)
x1, y1 = 0, 0
x2, y2 = 3, 4
jarak = math.sqrt((x2-x1)**2 + (y2-y1)**2)
print(f"Jarak (0,0) ke (3,4) = {jarak}")
```

## 3⃣ Konstanta Pi (π)

Konstanta π (pi) sangat penting dalam matematika, fisika, dan teknik. Python menyediakannya melalui pustaka `math`.

```python
# --- math.pi ---
import math

print(f"math.pi = {math.pi}")
print(f"pi s/d 10 desimal = {math.pi:.10f}")

# 5 desimal pertama: 3.14159
```

```python
# --- Aplikasi pi: Lingkaran ---
import math

r = 7
luas = math.pi * r ** 2            # pi r^2
keliling = 2 * math.pi * r         # 2 pi r

print(f"Jari-jari: {r}")
print(f"Luas lingkaran: {luas:.2f}")
print(f"Keliling lingkaran: {keliling:.2f}")

# Volume bola: V = 4/3 pi r^3
volume = 4/3 * math.pi * r ** 3
print(f"Volume bola: {volume:.2f}")
```

## 4⃣ Sinus & Cosinus

Fungsi trigonometri dasar — **sinus (sin)** dan **cosinus (cos)** — sangat penting dalam teknik, fisika, dan pengolahan sinyal.

> **Catatan:** Fungsi trigonometri di Python menggunakan **radian**, bukan derajat. Gunakan `math.radians()` untuk konversi.

```python
# --- Konversi Derajat <-> Radian ---
import math

derajat = 90
rad = math.radians(derajat)   # derajat -> radian
print(f"{derajat} derajat = {rad:.4f} rad")

rad2 = 1.5708
der = math.degrees(rad2)      # radian -> derajat
print(f"{rad2:.4f} rad = {der:.2f} derajat")

# Konversi umum
sudut_derajat = [0, 30, 45, 60, 90, 180, 360]
for d in sudut_derajat:
    print(f"  {d:3d} derajat = {math.radians(d):.4f} rad")
```

```python
# --- Sinus ---
import math

print("--- Sinus ---")
for d in [0, 30, 45, 60, 90, 180]:
    rad = math.radians(d)
    sin = math.sin(rad)
    print(f"  sin({d:3d}) = {sin:.4f}")
```

```python
# --- Cosinus ---
import math

print("--- Cosinus ---")
for d in [0, 30, 45, 60, 90, 180]:
    rad = math.radians(d)
    cos = math.cos(rad)
    print(f"  cos({d:3d}) = {cos:.4f}")
```

```python
# --- Aplikasi: Vektor Gaya ---
import math

# Gaya F dengan sudut theta terhadap sumbu-x
F = 100       # Newton
theta = 30    # derajat

# Komponen gaya
theta_rad = math.radians(theta)
Fx = F * math.cos(theta_rad)   # komponen sumbu-x
Fy = F * math.sin(theta_rad)   # komponen sumbu-y

print(f"Gaya {F} N pada sudut {theta} derajat:")
print(f"  Fx = {Fx:.2f} N")
print(f"  Fy = {Fy:.2f} N")

# Resultan: akar(Fx^2 + Fy^2) harus = F
resultan = math.sqrt(Fx**2 + Fy**2)
print(f"  Resultan = {resultan:.2f} N -- sudah sesuai")
```

```python
# --- Aplikasi: Gelombang sinusoidal ---
import math

# Simulasi gelombang: y(t) = A sin(2 pi f t + phi)
A = 5           # Amplitudo
frekuensi = 2   # Hz
omega = 2 * math.pi * frekuensi
phi = 0         # fase awal

print("Simulasi gelombang y(t) = 5 sin(4 pi t):")
for t in range(11):  # t = 0..10
    y = A * math.sin(omega * t/10 + phi)
    print(f"  t={t/10:.1f}s -> y = {y:6.2f}")
```

## Rangkuman

| Konsep | Fungsi / Cara | Contoh |
|--------|--------------|--------|
| **Kuadrat/Pangkat** | `**`, `pow()`, `math.pow()` | `5 ** 2 = 25` |
| **Akar** | `math.sqrt()`, `** 0.5` | `math.sqrt(25) = 5` |
| **Pi (π)** | `math.pi` | `3.14159...` |
| **Sinus** | `math.sin(rad)` | `sin(90°) = 1` |
| **Cosinus** | `math.cos(rad)` | `cos(0°) = 1` |
